## 2. 변수와 자료형

변수는 값을 담는 이름표입니다. `=` 로 값을 **할당**합니다.

| 자료형 | 예시 | 설명 |
|---|---|---|
| `int` | `10`, `-3` | 정수 |
| `float` | `3.14`, `0.5` | 실수 |
| `str` | `'AICE'`, `"서울"` | 문자열 |
| `bool` | `True`, `False` | 참/거짓 |
| `NoneType` | `None` | 값이 없음 |

`type()`으로 자료형을 확인할 수 있습니다.

````python
age = 25
height = 175.5
name = "홍길동"
is_student = True
nothing = None

print(type(age), type(height), type(name), type(is_student), type(nothing))
````

````python
# 형 변환: int(), float(), str(), bool()
print(int("100") + 1)     # 문자열 -> 정수
print(float("3.5") * 2)   # 문자열 -> 실수
print(str(2026) + "년")   # 정수 -> 문자열
print(int(3.99))          # 실수 -> 정수 (소수점 버림)
print(round(3.567, 1))    # 반올림 (시험 답안에서 자주 사용!)
print(round(2.5), round(3.5))  # 파이썬은 '짝수 반올림' 규칙: 2, 4
````

## 3. 연산자

| 구분 | 연산자 |
|---|---|
| 산술 | `+` `-` `*` `/`(나누기) `//`(몫) `%`(나머지) `**`(거듭제곱) |
| 비교 | `==` `!=` `>` `<` `>=` `<=` |
| 논리 | `and` `or` `not` |
| 할당 | `=` `+=` `-=` `*=` `/=` |
| 멤버십 | `in` `not in` |

````python
a, b = 17, 5
print("a + b  =", a + b)
print("a / b  =", a / b)
print("a // b =", a // b)
print("a % b  =", a % b)
print("a ** 2 =", a ** 2)
````

````python
x = 10
print(x > 5 and x < 20)   # True
print(x < 5 or x == 10)   # True
print(not x == 10)        # False
print('A' in 'AICE')      # True

x += 5                    # x = x + 5
print(x)
````

> ⚠️ **pandas와의 차이**: pandas에서 여러 조건을 묶을 때는 `and`, `or` 대신 **`&`, `|`** 를 쓰고, 각 조건을 **괄호로 감싸야** 합니다.
> 예) `df[(df['age'] > 30) & (df['sex'] == 'M')]` — 03번 노트북에서 다시 다룹니다.

## 4. 문자열 다루기

````python
s = "AICE Associate"
print(len(s))            # 길이
print(s[0], s[-1])       # 인덱싱: 첫 글자, 마지막 글자
print(s[0:4])            # 슬라이싱: 0~3번째
print(s.upper(), s.lower())
print(s.replace("Associate", "Basic"))
print(s.split(" "))      # 공백 기준으로 나누어 리스트로
print("  공백 제거  ".strip())
````

````python
# f-string: 문자열 안에 변수 값을 넣는 가장 편한 방법
model = "RandomForest"
acc = 0.87654
print(f"{model}의 정확도는 {acc}입니다.")
print(f"{model}의 정확도는 {acc:.3f}입니다.")   # 소수점 3자리
print(f"{model}의 정확도는 {acc:.1%}입니다.")   # 백분율
````

## 5. 조건문 (if / elif / else)

- 조건 뒤에 `:` 를 붙이고, 실행할 코드는 **들여쓰기(공백 4칸)** 합니다.

````python
score = 82

if score >= 80:
    print("합격 (AICE Associate는 80점 이상 합격)")
elif score >= 60:
    print("아쉽게 불합격")
else:
    print("더 공부가 필요해요")
````

````python
# 한 줄 조건식 (삼항 연산자)
result = "Pass" if score >= 80 else "Fail"
result
````

## 6. 반복문 (for / while)

````python
# range(시작, 끝, 간격) : 끝 값은 포함하지 않음
for i in range(1, 6):
    print(i, end=" ")
print()

for i in range(0, 10, 3):
    print(i, end=" ")
````

````python
# 리스트를 순회하기
models = ["DecisionTree", "RandomForest", "XGBoost"]
for m in models:
    print(f"{m} 모델 학습 중...")
````

````python
# break: 반복 중단 / continue: 이번 회차만 건너뜀
for n in range(1, 11):
    if n % 2 == 0:
        continue          # 짝수는 건너뜀
    if n > 7:
        break             # 7보다 크면 종료
    print(n, end=" ")
````

````python
# while: 조건이 참인 동안 반복
epoch = 0
loss = 1.0
while loss > 0.2:
    epoch += 1
    loss = loss * 0.7
    print(f"epoch {epoch}: loss={loss:.3f}")
````

## 7. 함수 (def, return, lambda)

````python
def bmi(weight, height_cm):
    """체중(kg)과 키(cm)로 BMI를 계산"""
    h = height_cm / 100
    return round(weight / h ** 2, 1)

print(bmi(70, 175))
````

````python
# 기본값(default) 매개변수와 키워드 인자
def greet(name, msg="안녕하세요"):
    return f"{name}님, {msg}!"

print(greet("홍길동"))
print(greet("홍길동", msg="합격을 축하합니다"))   # scikit-learn 함수들도 이런 식으로 옵션을 줍니다
````

````python
# lambda: 이름 없는 한 줄 함수 (pandas의 apply와 함께 자주 사용)
square = lambda x: x ** 2
print(square(7))

to_yn = lambda v: 1 if v == 'Y' else 0
print(to_yn('Y'), to_yn('N'))
````

## 9. 예외 처리 (try / except)

에러가 나도 프로그램이 멈추지 않게 합니다. 시험에서는 **에러 메시지를 읽고 원인을 고치는 능력**이 더 중요합니다.

| 자주 보는 에러 | 원인 |
|---|---|
| `NameError` | 정의하지 않은 변수/함수 이름 사용 (오타!) |
| `SyntaxError` | 문법 오류 (괄호, 콜론 누락 등) |
| `TypeError` | 자료형이 맞지 않음 (`'1' + 1`) |
| `KeyError` | 없는 컬럼명/딕셔너리 키 |
| `AttributeError` | 없는 메서드 호출 (`df.sort_index('col')` 처럼 잘못된 사용 포함) |
| `ValueError` | 값이 잘못됨 (`int('abc')`, 문자열을 모델에 넣음) |

````python
try:
    value = int("abc")
except ValueError as e:
    print("에러 발생:", e)
finally:
    print("finally 블록은 항상 실행됩니다")
````

## ✅ 정답 및 해설
> 먼저 직접 풀어본 뒤 확인하세요.

````python
# 문제 1
import numpy as np
import pandas as pd

# 문제 2
price = 12900
qty = 3
total = price * qty
print(f"총액: {total}원")

# 문제 3
답안3 = 0
for i in range(1, 101):
    if i % 3 == 0:
        답안3 += i
print(답안3)                                   # 1683
print(sum(range(3, 101, 3)))                  # 같은 결과를 한 줄로

# 문제 4
def grade(score):
    if score >= 90:
        return 'A'
    elif score >= 80:
        return 'B'
    elif score >= 70:
        return 'C'
    else:
        return 'F'
print(grade(85))

# 문제 5 : ① def 줄 끝 콜론(:) 누락  ② totl -> total 오타  ③ print 닫는 괄호 누락
def mean(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)

print(mean([80, 90, 100]))
````
