## 1. 리스트 (list)

- 대괄호 `[]` 로 만들며, **순서가 있고 수정 가능**합니다.
- 인덱스는 **0부터** 시작하고, 음수 인덱스는 뒤에서부터 셉니다.

````python
scores = [85, 92, 78, 95, 88]
print(scores[0], scores[-1])     # 첫 번째, 마지막
print(scores[1:3])               # 1, 2번 인덱스 (3은 미포함)
print(scores[:3])                # 처음부터 2번까지
print(scores[::2])               # 2칸씩 건너뛰며
print(len(scores), sum(scores), max(scores), min(scores))
````

````python
fruits = ['apple', 'banana']
fruits.append('cherry')          # 끝에 추가
fruits.insert(0, 'mango')        # 위치 지정 추가
fruits.extend(['kiwi', 'grape']) # 여러 개 추가
print(fruits)

fruits.remove('banana')          # 값으로 삭제
last = fruits.pop()              # 마지막 값 꺼내기
print(fruits, '/ 꺼낸 값:', last)

print('kiwi' in fruits)          # 포함 여부
print(fruits.index('cherry'))    # 위치 찾기
````

````python
nums = [5, 2, 9, 1, 7]
print(sorted(nums))                  # 정렬된 새 리스트 (원본 유지)
print(sorted(nums, reverse=True))    # 내림차순
nums.sort()                          # 원본 자체를 정렬
print(nums)

# 리스트 연산
print([1, 2] + [3, 4])
print([0] * 5)
````

## 2. 튜플 (tuple)

- 소괄호 `()` 로 만들며 **수정할 수 없습니다**.
- 여러 값을 한 번에 받는 **언패킹**에 자주 쓰입니다.

````python
shape = (891, 12)
rows, cols = shape                # 언패킹
print(f"행: {rows}, 열: {cols}")

# 함수가 여러 값을 반환하면 튜플로 반환됩니다
def min_max(values):
    return min(values), max(values)

lo, hi = min_max([3, 8, 1, 9])
print(lo, hi)

# train_test_split 도 4개의 값을 튜플로 반환합니다
# X_train, X_valid, y_train, y_valid = train_test_split(X, y, ...)
````

## 3. 딕셔너리 (dict)

- 중괄호 `{}` 와 `키: 값` 쌍으로 만듭니다.
- 키로 값을 찾습니다. (`dict[key]`)

````python
param_grid = {
    'max_depth': [3, 5, 7, 10],
    'min_samples_split': [2, 3, 5, 10]
}
print(param_grid['max_depth'])

param_grid['criterion'] = ['gini', 'entropy']   # 추가
print(param_grid.keys())
print(param_grid.values())

for k, v in param_grid.items():
    print(k, '->', v)
````

````python
# 존재하지 않는 키를 조회하면 KeyError → get()은 기본값 반환
d = {'Y': 1, 'N': 0}
print(d.get('Y'), d.get('X'), d.get('X', -1))

# 딕셔너리는 pandas에서 값 치환(map, replace)과 컬럼명 변경(rename)에 사용됩니다
# df['termination'].map({'Y': 1, 'N': 0})
# df.rename(columns={'avg_bill': '평균요금'})
````

````python
# 학습 기록(history)과 같은 구조 다루기
history = {'loss': [0.69, 0.52, 0.41, 0.37], 'val_loss': [0.66, 0.55, 0.47, 0.46]}
print("마지막 val_loss:", history['val_loss'][-1])
print("가장 낮은 val_loss:", min(history['val_loss']))
````

## 4. 집합 (set)

- 중복을 허용하지 않고 순서가 없습니다. **고유값 확인**에 유용합니다.

````python
cities = ['서울', '부산', '서울', '대구', '부산']
unique_cities = set(cities)
print(unique_cities, len(unique_cities))

a = {1, 2, 3, 4}
b = {3, 4, 5}
print(a & b, a | b, a - b)   # 교집합, 합집합, 차집합
````

## 5. 자주 쓰는 패턴

### 5-1. 리스트 컴프리헨션

````python
squares = [x ** 2 for x in range(1, 6)]
evens = [x for x in range(1, 11) if x % 2 == 0]
labels = ['Pass' if s >= 80 else 'Fail' for s in [85, 72, 90]]
print(squares, evens, labels)

# 딕셔너리 컴프리헨션
cols = ['age', 'sex', 'class']
col_map = {c: c.upper() for c in cols}
print(col_map)
````

### 5-2. enumerate, zip

````python
models = ['LogisticRegression', 'DecisionTree', 'RandomForest']
scores = [0.81, 0.78, 0.86]

for i, m in enumerate(models):
    print(i, m)

for m, s in zip(models, scores):
    print(f"{m:20s} {s:.2f}")

best = dict(zip(models, scores))
print(best)
print("최고 모델:", max(best, key=best.get))
````

## 6. numpy 기초

numpy는 숫자 계산을 빠르게 해주는 라이브러리로, pandas와 scikit-learn의 기반입니다.
스케일링(`StandardScaler`) 결과나 `model.predict()` 결과는 **numpy 배열(ndarray)** 입니다.

````python
import numpy as np

arr = np.array([1, 2, 3, 4, 5])
print(arr, type(arr))
print(arr * 10)            # 리스트와 달리 원소별 연산
print(arr.mean(), arr.std(), arr.sum())

mat = np.array([[1, 2, 3],
                [4, 5, 6]])
print(mat.shape)           # (행, 열)
print(mat[1, 2])           # 1행 2열
print(mat[:, 0])           # 모든 행의 0번째 열
````

````python
# 자주 쓰는 numpy 함수
print(np.max(mat), np.min(mat), np.argmax([3, 9, 2]))   # argmax: 최댓값의 위치
print(np.round(np.array([1.234, 5.678]), 1))
print(np.arange(0, 10, 2))
print(np.zeros(3), np.ones((2, 2)))
print(np.where(arr > 2, 'big', 'small'))   # 조건에 따라 값 선택
````

````python
# 예측 확률 -> 클래스 변환 (딥러닝에서 자주 사용)
proba = np.array([[0.9, 0.1],
                  [0.3, 0.7],
                  [0.6, 0.4]])
print(np.argmax(proba, axis=1))     # 각 행에서 가장 큰 값의 위치 -> [0 1 0]

p = np.array([0.12, 0.85, 0.51])
print((p > 0.5).astype(int))        # 임계값 0.5 기준 0/1 변환
````
