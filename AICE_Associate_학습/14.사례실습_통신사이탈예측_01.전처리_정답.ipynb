고객 기본정보(`cust_info.csv`)와 서비스 가입정보(`service_info.csv`)가 서로 다른 시스템에 저장되어 있어, 두 데이터를 **결합 → 정제 → 탐색 → 인코딩**한 뒤 파일로 저장합니다.<br>

`cust_info.csv` (고객 기본정보)

- gender : 성별 (Male/Female)
- SeniorCitizen : 고령자 여부 (1/0)

- Dependents : 부양가족 유무 (Yes/No)

- Churn : 이탈 여부 (Yes/No) ← **예측 대상**

`service_info.csv` (서비스 가입정보)
- customerID : 고객 ID
- PhoneService, MultipleLines : 전화 서비스, 다회선 여부
- InternetService : 인터넷 종류 (DSL / Fiber optic / No)
- OnlineSecurity, OnlineBackup, DeviceProtection, TechSupport, StreamingTV, StreamingMovies : 부가서비스 가입 여부
- Contract : 계약 형태 (Month-to-month / One year / Two year)
- PaperlessBilling : 종이없는 청구서 여부
- PaymentMethod : 결제 수단
- MonthlyCharges : 월 요금
- TotalCharges : 누적 요금

### **5. TotalCharges(누적 요금)는 숫자여야 하는데 자료형이 object입니다. 원인을 찾아 숫자형으로 바꾸세요.**
---
- df.info()로 자료형을 확인하세요.
- TotalCharges 값 중 공백 문자열(' ')의 개수를 '답안5' 변수에 저장하세요.
- 공백 문자열을 결측치(np.nan)로 바꾼 뒤 TotalCharges를 float 자료형으로 변환하세요.
- 아래 코드의 오류를 정정하세요.
---

```python
# [5번 정답]
df.info()

print((df['TotalCharges'] == ' ').sum())
답안5 = 11

df['TotalCharges'] = df['TotalCharges'].replace(' ', np.nan)
df['TotalCharges'] = df['TotalCharges'].astype(float)
print(df['TotalCharges'].dtype)

# 참고: pd.to_numeric(df['TotalCharges'], errors='coerce') 로 한 번에 처리할 수도 있습니다.
```

<br>

### **6. 결측치를 처리하세요.**

- **[6-1] df의 컬럼별 결측치 개수를 확인하고, 결측치가 있는 컬럼 수를 '답안6_1'에, 전체 결측치 개수를 '답안6_2'에 저장하세요.**
---

```python
# [6-1번 정답]
na = df.isnull().sum()
print(na[na > 0])
답안6_1 = 4
답안6_2 = 84
```

- **[6-2] 아래 가이드에 따라 결측치를 처리하고, 결과를 df_na에 저장하세요.**
    - TotalCharges : 결측인 고객은 모두 가입 기간(tenure)이 0개월이므로 **0** 으로 채우세요.
    - MonthlyCharges : **중앙값**으로 채우세요.
    - Dependents, PaymentMethod : 각 컬럼의 **최빈값**으로 채우세요.
    - 처리 후 결측치가 없는지 확인하세요.
---

```python
# [6-2번 정답]
print(df.loc[df['TotalCharges'].isnull(), 'tenure'].unique())     # 정말 tenure가 0인지 확인

df_na = df.copy()
df_na['TotalCharges'] = df_na['TotalCharges'].fillna(0)
df_na['MonthlyCharges'] = df_na['MonthlyCharges'].fillna(df_na['MonthlyCharges'].median())
for col in ['Dependents', 'PaymentMethod']:
    df_na[col] = df_na[col].fillna(df_na[col].mode()[0])

print(df_na.isnull().sum().sum())
```

### **7. 이탈 여부(Churn)의 분포와 가입 기간(tenure)과의 관계를 시각화하세요.**
---
- 대상 데이터프레임 : df_na
- subplots로 1행 2열 그래프를 그리세요. (figsize=(14, 5))
- 왼쪽: Churn의 countplot, 오른쪽: Churn별로 구분(hue)한 tenure의 histplot
- 제목은 각각 'Churn Count', 'Tenure by Churn'
- 이탈 고객(Yes)의 비율을 **소수점 둘째 자리**까지 반올림해 '답안7'에 저장하세요. (예: 답안7 = 0.12)
---

```python
# [7번 정답]
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))
sns.countplot(data=df_na, x='Churn', ax=axes[0])
sns.histplot(data=df_na, x='tenure', hue='Churn', ax=axes[1])
axes[0].set_title('Churn Count')
axes[1].set_title('Tenure by Churn')
plt.show()

print(df_na['Churn'].value_counts(normalize=True))
답안7 = 0.35
```

> 💡 **해석**: 가입 초기(tenure가 짧은) 고객의 이탈이 많습니다.

### **8. 계약 형태(Contract)별 이탈률을 비교하세요.**
---
- Contract별 이탈 비율(Churn이 'Yes'인 비율)을 구하세요. (crosstab 또는 groupby 활용)
- seaborn countplot으로 Contract별 Churn 분포를 그리세요. (x='Contract', hue='Churn')
- 이탈률이 가장 높은 계약 형태를 '답안8'에 저장하세요.

### **9. 월 요금(MonthlyCharges)과 인터넷 서비스(InternetService)가 이탈과 어떤 관계가 있는지 확인하세요.**
---
- 왼쪽: x='Churn', y='MonthlyCharges' 인 boxplot
- 오른쪽: x='InternetService', y='Churn_num' 인 barplot (Churn_num은 Churn을 Yes=1, No=0으로 바꾼 임시 컬럼)
- subplots(1, 2, figsize=(14, 5)) 사용
- 이탈률이 가장 높은 InternetService 값을 '답안9'에 저장하세요.
---

```python
# [9번 정답]
tmp = df_na.copy()
tmp['Churn_num'] = tmp['Churn'].map({'Yes': 1, 'No': 0})

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=tmp, x='Churn', y='MonthlyCharges', ax=axes[0])
sns.barplot(data=tmp, x='InternetService', y='Churn_num', ax=axes[1])
plt.show()

print(tmp.groupby('InternetService')['Churn_num'].mean())
답안9 = 'Fiber optic'
```

### **11. 모델 학습을 위해 데이터를 인코딩하세요.**
---
- 대상 데이터프레임 : df_na
- 모델링에 불필요한 customerID 컬럼을 삭제하세요.
- Churn 컬럼은 'Yes'는 1, 'No'는 0으로 변환하세요.
- 나머지 object형(문자형) 컬럼 전체를 원-핫 인코딩하세요. (dtype=int 옵션으로 0/1 정수로)
- 결과는 df_preset에 저장하고, df_preset의 컬럼 개수를 '답안11'에 저장하세요.

```python
# [11번 정답]
df_preset = df_na.drop('customerID', axis=1)
df_preset['Churn'] = df_preset['Churn'].map({'Yes': 1, 'No': 0})

obj_cols = df_preset.select_dtypes(include='object').columns.tolist()
print(obj_cols)
df_preset = pd.get_dummies(data=df_preset, columns=obj_cols, dtype=int)
df_preset.info()

답안11 = 46
```
