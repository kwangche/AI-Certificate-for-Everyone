## 6. 행 선택하기: `loc` vs `iloc`

| | `loc` | `iloc` |
|---|---|---|
| 기준 | 인덱스 **이름(label)** | 인덱스 **위치(정수)** |
| 슬라이싱 끝 | **포함** | 미포함 |
| 예 | `df.loc[0:2, 'age']` | `df.iloc[0:2, 2]` |

````python
df.loc[0:2, ['class', 'age']]     # 0, 1, 2행 (끝 포함)
````

````python
df.iloc[0:2, 0:3]                 # 0, 1행 / 0~2열 (끝 미포함)
````

````python
df.iloc[-1]                       # 마지막 행
````

````python
df.loc[5, 'avg_bill']             # 특정 셀 하나
````

## 7. 조건으로 필터링하기 (Boolean Indexing)

- 조건식은 True/False로 이루어진 Series를 만듭니다. 이를 `df[ ]` 에 넣으면 True인 행만 남습니다.
- 여러 조건: `&`(and), `|`(or), `~`(not) + **각 조건을 괄호로 감싸기**

````python
cond = df['age'] >= 60
cond.head()
````

````python
df[cond].head()
````

````python
# 60세 이상이면서 해지한 고객
df[(df['age'] >= 60) & (df['termination'] == 'Y')].shape
````

````python
# 미납이거나 일시정지한 고객
df[(df['npay'] == 'Y') | (df['stop'] == 'Y')].shape
````

````python
# isin: 여러 값 중 하나에 해당
df[df['class'].isin(['A', 'B'])].head(3)
````

````python
# ~ : 조건 반대
df[~df['class'].isin(['A', 'B'])]['class'].unique()
````

````python
# between: 범위
df[df['age'].between(30, 39)].shape
````

````python
# query: 문자열로 조건 작성
df.query("age >= 60 and termination == 'Y'").shape
````

````python
# loc에 조건과 컬럼을 함께
df.loc[df['avg_bill'] > 50000, ['class', 'avg_bill']].head()
````

> ⚠️ 샘플 문항 11번에서 `sort_index('importance', ...)` 를 `sort_values('importance', ...)` 로 고치는 문제가 나왔습니다. 두 함수를 꼭 구분하세요.

````python
df.sort_values('avg_bill', ascending=False).head()
````

````python
df.sort_values(['class', 'age'], ascending=[True, False]).head()
````

````python
df.nlargest(3, 'avg_bill')
````
